```python
from pathlib import Path
from importlib import metadata
import platform
import sys

# Leave blank unless the project folder cannot be found.
PROJECT_ROOT_OVERRIDE = ""

def is_project_root(candidate):
    return (candidate.name == "MLOP-PJ"
            and (candidate / "README.md").is_file()
            and (candidate / "proposals" / "current").is_dir())

working_dir = Path.cwd().resolve()
candidates = []
if PROJECT_ROOT_OVERRIDE.strip():
    candidates.append(Path(PROJECT_ROOT_OVERRIDE).expanduser().resolve())
else:
    for base in [working_dir, *working_dir.parents]:
        candidates.extend([base, base / "MLOP-PJ",
                           base / "course" / "MLOP-PJ",
                           base / "MLop" / "public_teaching_mlaiops" / "course" / "MLOP-PJ"])
PROJECT_ROOT = next((p for p in candidates if is_project_root(p)), None)

is_wsl = platform.system() == "Linux" and "microsoft" in platform.release().lower()
print("Python version :", sys.version.split()[0])
print("Python kernel  :", sys.executable)
print("System         :", "WSL" if is_wsl else platform.system())
print("Working folder :", working_dir)
print("Project folder :", PROJECT_ROOT if PROJECT_ROOT else "NOT FOUND")
print("\nInstalled distribution metadata (not compatibility tests):")
for distribution in ["ipykernel", "numpy", "pandas", "scikit-learn", "matplotlib", "mlflow", "dvc"]:
    try:
        version = metadata.version(distribution)
    except metadata.PackageNotFoundError:
        version = "NOT INSTALLED"
    print(f"  {distribution:14} {version}")

if PROJECT_ROOT is None:
    print("\nCheckpoint: send this output; project path needs to be set before data work.")
else:
    print("\nCheckpoint: send this output and wait for Step 2.")
```
